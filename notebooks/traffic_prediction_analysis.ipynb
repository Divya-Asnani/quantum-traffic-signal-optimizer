```python
df["DateTime"] = pd.to_datetime(df["DateTime"])

print(df["DateTime"].diff().value_counts().head(10))
```

```python
df = df.sort_values(["Junction", "DateTime"]).reset_index(drop=True)

print(df.head(10))
```

```python
time_diff = df.groupby("Junction")["DateTime"].diff()

print(time_diff.value_counts().head(10))
```

```python
#feature engineering
# hour
# day_of_week
# month
# is_weekend
```

```python
df["hour"] = df["DateTime"].dt.hour

df["day_of_week"] = df["DateTime"].dt.dayofweek

df["month"] = df["DateTime"].dt.month

df["is_weekend"] = (df["day_of_week"] >= 5).astype(int)
```

```python
df["traffic_1h_ago"] = (
    df.groupby("Junction")["Vehicles"].shift(1)
)

df["traffic_2h_ago"] = (
    df.groupby("Junction")["Vehicles"].shift(2)
)

df["traffic_3h_ago"] = (
    df.groupby("Junction")["Vehicles"].shift(3)
)
```

```python
df["rolling_mean_3h"] = (
    df.groupby("Junction")["Vehicles"]
      .shift(1)
      .rolling(3)
      .mean()
      .reset_index(level=0, drop=True)
)
```

```python
df["target_vehicles"] = (
    df.groupby("Junction")["Vehicles"].shift(-1)
)

print(df[[
    "DateTime",
    "Junction",
    "Vehicles",
    "traffic_1h_ago",
    "traffic_2h_ago",
    "traffic_3h_ago",
    "target_vehicles"
]].head(10))
```

```python
print("Target missing values:", df["target_vehicles"].isnull().sum())
```

```python
df_model = df.dropna(
    subset=[
        "traffic_1h_ago",
        "traffic_2h_ago",
        "traffic_3h_ago",
        "rolling_mean_3h",
        "target_vehicles"
    ]
).copy()
```

```python
print("Original rows:", len(df))
print("Rows available for modeling:", len(df_model))
print("Rows removed:", len(df) - len(df_model))
```

```python
# Make sure the modeling data is ordered by time
df_model = df_model.sort_values("DateTime").reset_index(drop=True)

# Find the 80% time point
split_index = int(len(df_model) * 0.80)
split_time = df_model.loc[split_index, "DateTime"]

# Split chronologically
train_df = df_model[df_model["DateTime"] < split_time].copy()
test_df = df_model[df_model["DateTime"] >= split_time].copy()

print("Split time:", split_time)

print("\nTraining set:")
print("Rows:", len(train_df))
print("Start:", train_df["DateTime"].min())
print("End:", train_df["DateTime"].max())

print("\nTest set:")
print("Rows:", len(test_df))
print("Start:", test_df["DateTime"].min())
print("End:", test_df["DateTime"].max())
```

```python
print("Training junctions:")
print(train_df["Junction"].value_counts())

print("\nTesting junctions:")
print(test_df["Junction"].value_counts())
```

```python
feature_columns = [
    "hour",
    "day_of_week",
    "month",
    "is_weekend",
    "Junction",
    "traffic_1h_ago",
    "traffic_2h_ago",
    "traffic_3h_ago",
    "rolling_mean_3h"
]

target_column = "target_vehicles"

X_train = train_df[feature_columns]
y_train = train_df[target_column]

X_test = test_df[feature_columns]
y_test = test_df[target_column]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```
